```python
age = int(input("Enter your age: "))
monthly_income = int(input("Enter your monthly income: "))

if age >= 18 and monthly_income >= 20000 : 
    print("You meet the criteria.")
else: 
    print("You do not meet the criteria.")
```

```python
mark = int(input("Mark: "))

if mark >= 80:
    print("A+")
elif mark >= 70:
    print("A")
elif mark >= 60:
    print("A-")
elif mark >= 50:
    print("B")
elif mark >= 40:
    print("C")
else:
    print("F")
```
